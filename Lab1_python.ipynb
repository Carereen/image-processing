```python
import cv2
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt
import os
from os import listdir

path = '/content/drive/My Drive/Lab1-Image/'
i = 1

for img in os.listdir(path):
  if img.endswith((".jpg", "JPG", "bmp", "BMP", "tiff", "TIFF")):
      img_show = cv2.imread(path + img)

      plt.subplot(2, 2, i)
      plt.title(img)
      plt.xticks([]), plt.yticks([])
      plt.imshow(img_show)

      i = i + 1

plt.show()
```

```python
path = '/content/drive/My Drive/Lab1-Image/'
path_new = '/content/drive/My Drive/'

image = cv2.imread(path + "image_dog.JPG")
image2 = image

cv2.imwrite(path_new+"image_dog_new.jpg", image2)
image_new = cv2.imread(path_new + 'image_dog_new.jpg')

plt.title("Image: image_dog_new.jpg")
plt.imshow(image_new)
plt.show()
```
