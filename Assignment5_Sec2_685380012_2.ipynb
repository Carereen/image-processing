```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import sys
import os
from math import exp, sqrt
```

```python
# task 1
path = "/content/drive/MyDrive/2568-term2/image/Lab5/"
image_name = "img1.jpg"
image_bgr = cv2.imread(path + image_name)

if image_bgr is None:
  print(f"Image file not found. Please check the path: {path + image_name} .")
  sys.exit()

plt.figure(figsize = (15, 10), constrained_layout = False)

plt.subplot(131), plt.imshow(image_bgr)
plt.title(image_name + ": BGR")

image1 = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2HLS)
plt.subplot(132), plt.imshow(image1)
plt.title(image_name + ": BGR TO HLS")

image2 = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2Lab)
plt.subplot(133), plt.imshow(image2)
plt.title(image_name + ": BGR TO Lab")
plt.show()

plt.figure(figsize = (15, 10), constrained_layout = False)
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
plt.subplot(131), plt.imshow(image_rgb)
plt.title(image_name + ": RGB")

image3 = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HLS)
plt.subplot(132), plt.imshow(image3)
plt.title(image_name + ": RGB TO HLS")

image3 = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2Lab)
plt.subplot(133), plt.imshow(image3)
plt.title(image_name + ": RGB TO Lab")
plt.show()
```

```python
# task 2
path = "/content/drive/MyDrive/2568-term2/image/Lab5/"
image_name = "img1.jpg"
image_bgr = cv2.imread(path + image_name)

if image_bgr is None:
  print(f"Image file not found. Please check the path: {path + image_name} .")
  sys.exit()

plt.figure(figsize = (15, 10), constrained_layout = False)

image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
plt.subplot(121), plt.imshow(image_rgb)
plt.title("RGB image")

blur_bgr = cv2.GaussianBlur(image_bgr, (5, 5), 0)
image_rgb_blur = cv2.cvtColor(blur_bgr, cv2.COLOR_BGR2RGB)
plt.subplot(122), plt.imshow(image_rgb_blur)
plt.title("RGB GaussianBlur image")
plt.show()

plt.figure(figsize = (15, 10), constrained_layout = False)

image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
plt.subplot(121), plt.imshow(image_rgb)
plt.title("RGB image")

hsv = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2HSV)
hsv_blur = cv2.GaussianBlur(hsv, (5, 5), 0)
hsv_blur_rgb = cv2.cvtColor(hsv_blur, cv2.COLOR_HSV2RGB)
plt.subplot(122), plt.imshow(hsv_blur_rgb)
plt.title("HSV GaussianBlur image")
plt.show()
```

```python
# task 3
path = "/content/drive/MyDrive/2568-term2/image/Lab5/"
image_name = "color_circle.jpg"
image_bgr = cv2.imread(path + image_name)

if image_bgr is None:
  print(f"Image file not found. Please check the path: {path + image_name} .")
  sys.exit()

image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
image_hsv = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2HSV)

# Red
lower_range1 = np.array([round(0/ 2), 50, 50])
upper_range1 = np.array([round(15 / 2), 255, 255])

lower_range2 = np.array([round(340 / 2), 50, 50])
upper_range2 = np.array([round(358 / 2), 255, 255])

mask1 = cv2.inRange(image_hsv, lower_range1, upper_range1)
mask2 = cv2.inRange(image_hsv, lower_range2, upper_range2)

mask_red = mask1 | mask2

plt.figure(figsize = (15, 10), constrained_layout = False)
result_red = cv2.bitwise_and(image_rgb, image_rgb, mask = mask_red)
plt.subplot(131), plt.imshow(result_red)
plt.title("Red Detection")

# Orange
lower_range = np.array([round(15 / 2), 50, 50])
upper_range = np.array([round(30 / 2), 255, 255])

mask_orange = cv2.inRange(image_hsv, lower_range, upper_range)
result_orange = cv2.bitwise_and(image_rgb, image_rgb, mask = mask_orange)
plt.subplot(132), plt.imshow(result_orange)
plt.title("Orange Detection")

#Blue
lower_range = np.array([round(45 / 2), 50, 50])
upper_range = np.array([round(60 / 2), 255, 255])

mask_yellow = cv2.inRange(image_hsv, lower_range, upper_range)
result_yellow = cv2.bitwise_and(image_rgb, image_rgb, mask = mask_yellow)
plt.subplot(133), plt.imshow(result_yellow)
plt.title("Blue Detection")
plt.show()

plt.figure(figsize = (15, 10), constrained_layout = False)

# violet
lower_range = np.array([round(285 / 2), 50, 50])
upper_range = np.array([round(315 / 2), 255, 255])

mask_violet = cv2.inRange(image_hsv, lower_range, upper_range)
result_violet = cv2.bitwise_and(image_rgb, image_rgb, mask = mask_violet)
plt.subplot(131), plt.imshow(result_violet)
plt.title("violet Detection")

# blue
lower_range = np.array([round(210 / 2), 50, 50])
upper_range = np.array([round(270 / 2), 255, 255])

mask_blue = cv2.inRange(image_hsv, lower_range, upper_range)
result_blue = cv2.bitwise_and(image_rgb, image_rgb, mask = mask_blue)
plt.subplot(132), plt.imshow(result_blue)
plt.title("Blue Detection")

# green
lower_range = np.array([round(120 / 2), 50, 50])
upper_range = np.array([round(150 / 2), 255, 255])

mask_green = cv2.inRange(image_hsv, lower_range, upper_range)
result_green = cv2.bitwise_and(image_rgb, image_rgb, mask = mask_green)
plt.subplot(133), plt.imshow(result_green)
plt.title("Orange Detection")
plt.show()
```
