```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import sys
import os
```

```python
# task 2
path = "/content/drive/MyDrive/2568-term2/image/Lab5/"
image_name = "cat_newyear.jpg"
image_bgr = cv2.imread(path + image_name)

if image_bgr is None:
  print(f"Image file not found. Please check the path: {path + image_name} .")
  sys.exit()

image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize = (15, 10), constrained_layout = False)
plt.subplot(221)
plt.title("RGB Image")
plt.imshow(image_rgb)

color = ('r', 'g', 'b')
plt.subplot(223)
plt.title("Histogram of R, G and B Chanel")

for i, col in enumerate(color):
  histr = cv2.calcHist([image_rgb], [i], None, [256], [0, 256])
  plt.plot(histr, color = col)
  plt.xlim([0, 256])

image_new = image_rgb.copy()

image_new[:, :, 0] = cv2.equalizeHist(image_rgb[:, :, 0])
image_new[:, :, 1] = cv2.equalizeHist(image_rgb[:, :, 1])
image_new[:, :, 2] = cv2.equalizeHist(image_rgb[:, :, 2])
plt.subplot(222)
plt.title("RGB Equalize Histogram Image")
plt.imshow(image_new)

plt.subplot(224)
color = ('r', 'g', 'b')
plt.title("Histogram of R, G and B Chanel on RGB Equalize Histogram Image")

for i, col in enumerate(color):
  histr_new = cv2.calcHist([image_new], [i], None, [256], [0, 256])
  plt.plot(histr_new, color = col)
  plt.xlim([0, 256])

plt.show()
```

```python
# task 4
path = "/content/drive/MyDrive/2568-term2/image/Lab5/"
image_name = "cat_newyear.jpg"
image_bgr = cv2.imread(path + image_name)

if image_bgr is None:
  print(f"Image file not found. Please check the path: {path + image_name} .")
  sys.exit()

image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize = (15, 10), constrained_layout = False)
plt.subplot(131)
plt.title("RGB Image")
plt.imshow(image_rgb)

image_new = image_rgb. copy()
kernel_3x3 = np.ones((3, 3), np.float32) / 9

image_new[:, :, 0] = cv2.filter2D(image_new[:, :, 0], -1, kernel_3x3)
image_new[:, :, 1] = cv2.filter2D(image_new[:, :, 1], -1, kernel_3x3)
image_new[:, :, 2] = cv2.filter2D(image_new[:, :, 2], -1, kernel_3x3)

plt.subplot(132)
plt.title("RGB Smoothing Image (kernel:3x3)")
plt.imshow(image_new)

kernel_7x7 = np.ones((7,7), np.float32) / 49
image_smooth_7 = cv2.filter2D(image_rgb, -1, kernel_7x7)

plt.subplot(133),
plt.title("RGB Smoothing Image (kernel: 7x7)")
plt.imshow(image_smooth_7)

plt.show()
```

```python
# task 6
path = "/content/drive/MyDrive/2568-term2/image/Lab5/"
image_name = "color_circle.jpg"
image_bgr = cv2.imread(path + image_name)

if image_bgr is None:
  print(f"Image file not found. Please check the path: {path + image_name} .")
  sys.exit()

image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
image_hsv = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2HSV)

plt.figure(figsize = (15, 10), constrained_layout = False)
plt.subplot(121), plt.imshow(image_rgb)
plt.title(image_name + ": RGB.")
plt.subplot(122), plt.imshow(image_hsv)
plt.title(image_name + ": HSV.")
plt.show()

plt.figure(figsize = (15, 10), constrained_layout = False)

lower_range = np.array([round(210 / 2), 50, 50])
upper_range = np.array([round(270 / 2), 255, 255])

mask_blue = cv2.inRange(image_hsv, lower_range, upper_range)
result_blue = cv2.bitwise_and(image_rgb, image_rgb, mask = mask_blue)
plt.subplot(121), plt.imshow(result_blue)
plt.title("Blue Detection")

lower_range = np.array([round(15 / 2), 50, 50])
upper_range = np.array([round(30 / 2), 255, 255])

mask_orange = cv2.inRange(image_hsv, lower_range, upper_range)
result_orange = cv2.bitwise_and(image_rgb, image_rgb, mask = mask_orange)
plt.subplot(122), plt.imshow(result_orange)
plt.title("Orange Detection")
plt.show()
```

```python
# task 7
path = "/content/drive/MyDrive/2568-term2/image/Lab5/Nemo/"

if not os.path.exists(path):
  print(f"Path not found: {path}")
  sys.exit()

for img in os.listdir(path):
  if img.endswith(("jpg", "JPG", "bmp", "BMP", "tiff", "TIFF")):
    image = cv2.imread(path + img)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image_hsv = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HSV)

# array([H, S, V]) | s = Saturation (ความเข้มของสี) | v = Value (ความสว่าง)
lower_range = np.array([round(1 / 2), 50, 50])
upper_range = np.array([round(30 / 2), 255, 255])

mask_orange = cv2.inRange(image_hsv, lower_range, upper_range)
result_orange = cv2.bitwise_and(image_rgb, image_rgb, mask = mask_orange)

plt.figure(figsize = (15, 10), constrained_layout = False)
plt.subplot(121), plt.imshow(image_rgb)
plt.title(img)
plt.subplot(122), plt.imshow(result_orange)
plt.title("Orange Detection")
plt.show()
```
