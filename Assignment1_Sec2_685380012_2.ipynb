```python
import cv2
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt
import os
from os import listdir

path = '/content/drive/My Drive/image-assignment1/'

i = 1

for img in os.listdir(path):
  if img.endswith((".jpg", "JPG", ".png","bmp", "BMP", "tiff", "TIFF")):
    img_show = cv2.imread(path + img)

    plt.subplot(3,3, i)
    plt.title(img)
    plt.xticks([]), plt.yticks([])
    plt.imshow(img_show)

    i = i + 1

plt.show()
```

```python
import os
import cv2
import matplotlib.pyplot as plt

path = '/content/drive/My Drive/image-assignment1/'
path_new = '/content/drive/My Drive/'

images = []
i = 1

for img in os.listdir(path):
    if img.lower().endswith((".jpg", ".png", ".bmp", ".tiff")):
        old_path = os.path.join(path, img)

        new_name = f"{i}.png"
        new_path = os.path.join(path_new, new_name)

        image = cv2.imread(old_path)
        cv2.imwrite(new_path, image)


        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        images.append((new_name, image))

        i = i + 1

for idx, (name, img) in enumerate(images, start=1):
    plt.subplot(2, 3, idx)
    plt.imshow(img)
    plt.xticks([]), plt.yticks([])
    plt.title(name)

plt.show()
```
