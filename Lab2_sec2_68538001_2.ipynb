```python
image = cv2.imread(path + image_name, 1)
img_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

(h, w, d) = image_rgb.shape

r = 300 / w

dim = (300, int(h * r))

img_resize = cv2.resize(img_rgb, dim)

plt.imshow(img_resize)
plt.title("Resize Image (Ratio Determination)")
plt.show()
```

```python
import imutils
path = '/content/drive/My Drive/2568-term2/image/week2/'
image_name = 'cat_newyear.jpg'
image = cv2.imread(path + image_name, 1)
img_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

img_resize = imutils.resize(img_rgb, width = 300)

plt.imshow(img_resize)
plt.title("Resize Image using imutils.resize")
plt.show()
```

```python
# Task 7
img_rotated = imutils.rotate(img_rgb, -45)
plt.imshow(img_rotated), plt.title("rotated image -45")
plt.show()

img_rotated_bound = imutils.rotate_bound(img_rgb, 45)
plt.imshow(img_rotated_bound), plt.title("rotated_bound image 45")
plt.show()
```

```python
# rectangle
img_rectangle = img_rgb.copy()

cv2.rectangle(img_rectangle, (700, 200), (800, 350), (0, 255, 0), 3)

plt.imshow(img_rectangle)
plt.title("Rectangle Drawing")
plt.axis("off")
plt.show()
```

```python
# task 8
img_circle = img_rgb.copy()

cv2.circle(img_circle, (780, 270), 50, (255, 0, 0), 3)

plt.imshow(img_circle)
plt.title("Circle Drawing")
plt.axis("off")
plt.show()
```

```python
# task 9
img_line = img_rgb.copy()

cv2.line(img_line, (500, 350), (600, 250), (0, 0, 255), 4)

plt.imshow(img_line)
plt.title("Line Drawing")
plt.axis("off")
plt.show()
```
