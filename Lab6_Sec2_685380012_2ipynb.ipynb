```python
# task 6 - Morphological Gradient
image_name = 'img1.jpg'
image_gray = cv2.imread(path + image_name)

if image_gray is None:
  print(f"Image file not found. Please check the path: {path + image_name} .")
  sys.exit()

kernel = np.ones((7, 7), np.uint8)

image_gradient = cv2.morphologyEx(image_gray, cv2.MORPH_GRADIENT, kernel)

fig = plt.figure(figsize = (20, 20))

plt.subplot(1, 2, 1), plt.imshow(image_gray)
plt.title(image_name + ': gray image')

plt.subplot(1, 2, 2), plt.imshow(image_gradient)
plt.title(image_name + ': gradient image')
plt.show()
```

```python
# task 7 - Hole Filling
image_name = 'coin.jpg'
image_color = cv2.imread(path + image_name)

if image_color is None:
  print(f"Image file not found. Please check the path: {path + image_name} .")
  sys.exit()

image_rgb = cv2.cvtColor(image_color, cv2.COLOR_BGR2RGB)
image_gray = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2GRAY)

fig = plt.figure(figsize = (15, 15))
plt.subplot(2, 2, 1), plt.imshow(image_gray, cmap = 'gray')
plt.title(image_name + ': gray image')

# Threshold
# Set values equal to or above 240 to 0
# Set values below 240 to 255
th, image_th = cv2.threshold(image_gray, 240, 255, cv2.THRESH_BINARY_INV);

plt.subplot(2, 2, 2), plt.imshow(image_th, cmap = 'gray')
plt.title(image_name + ': threshold image')

image_floodfill = image_th.copy()

# เพราะ OpenCV กำหนดว่า mask ต้องใหญ่กว่าภาพ 2 pixel
h, w = image_th.shape[: 2]
mask = np.zeros((h + 2, w + 2), np.uint8)

cv2.floodFill(image_floodfill, mask, (0, 0), 255);

image_floodfill_inv = cv2.bitwise_not(image_floodfill)

plt.subplot(2, 2, 3), plt.imshow(image_floodfill_inv, cmap = 'gray')
plt.title(image_name + ': Invert flooddilled image')

image_out = cv2.bitwise_or(image_th, image_floodfill_inv)

plt.subplot(2, 2, 4), plt.imshow(image_out, cmap = 'gray')
plt.title(image_name + ': result image')
plt.show()
```

```python
# task 8 - Skeletonization
image_name = 'img1.jpg'
image = cv2.imread(path + image_name, 1)

if image is None:
  print(f"Image file not found. Please check the path: {path + image_name} .")
  sys.exit()

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image_gray = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2GRAY)

ret, img = cv2.threshold(image_gray, 127, 255, 0)
size = np.size(img)
skel = np.zeros(img.shape, np.uint8)

element = cv2.getStructuringElement(cv2.MORPH_CROSS, (3, 3))

while True:
  open = cv2.morphologyEx(img, cv2.MORPH_OPEN, element)
  temp = cv2.subtract(img, open)
  eroded = cv2.erode(img, element)
  skel = cv2.bitwise_or(skel, temp)
  img = eroded.copy()
  if cv2.countNonZero(img) == 0:
    break

fig = plt.figure(figsize = (15, 15))

plt.subplot(1, 2, 1), plt.imshow(image_gray, cmap = 'gray')
plt.title(image_name + ': gray image')

plt.subplot(1, 2, 2), plt.imshow(skel, cmap = 'gray')
plt.title(image_name + ': Skeletonization image')
plt.show()
```

```python
# task 9 - skimage
from skimage.morphology import skeletonize, thin
image_name = 'horse.jpg'
img = cv2.imread(path + image_name, 0)

if img is None:
  print(f"Image file not found. Please check the path: {path + image_name} .")
  sys.exit()

image = img > 200
skeleton = skeletonize(image)
thin = thin(image)

fig = plt.figure(figsize = (15, 10))

plt.subplot(2, 2, 1), plt.imshow(image, cmap = 'gray')
plt.title(image_name + ': gray image')

plt.subplot(2, 2, 2), plt.imshow(skeleton, cmap = 'gray')
plt.title(image_name + ': Skeletonization image')

plt.subplot(2, 2, 3), plt.imshow(thin, cmap = 'gray')
plt.title(image_name + ': thin image')
fig.tight_layout()
plt.show()
```
