```python
import cv2
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt
```

```python
# Task 1
import numpy as np

image = np.ones([8, 7, 3], np.uint8) * 255

image[1:7, 2:3, 0] = 0
image[1, 2:6, 0] = 0
image[3, 2:5, 0] = 0

image[1:7, 2:3, 1] = 255
image[1, 2:6, 1] = 255
image[3, 2:5, 1] = 255

image[1:7, 2:3, 2] = 0
image[1, 2:6, 2] = 0
image[3, 2:5, 2] = 0

plt.figure(figsize=(4,4))
plt.imshow(image)
plt.title("F Letter")
plt.show()
```

```python
# Task 2
import imutils
path = '/content/drive/My Drive/2568-term2/image/week2/'
image_name = 'cat_dog.jpg'
image = cv2.imread(path + image_name, 1)
img_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

img_rotated_bound = imutils.rotate_bound(img_rgb, 100)

plt.imshow(img_rgb)
plt.show()

plt.imshow(img_rotated_bound), plt.title("rotated_bound image 100")
plt.show()
```

```python
# Task 3
image = cv2.imread(path + image_name, 1)
img_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

(h, w, d) = img_rgb.shape

r = 2000 / w

dim = (2000, int(h * r))

img_resize = cv2.resize(img_rgb, dim)

plt.imshow(img_resize)
plt.title("Resize Image (Ratio Determination)")
plt.show()
```

```python
# Task 4
img_rectangle_circle = img_rgb.copy()

cv2.rectangle(img_rectangle_circle, (50, 30), (500, 450), (255, 0, 0), 3)
cv2.circle(img_rectangle_circle, (710, 650), 160, (0, 255, 0), 2)

cv2.putText(img_rectangle_circle, "Cat & Dog", (600,200), cv2.FONT_HERSHEY_SIMPLEX, 2, (0, 0, 255), 3)

plt.imshow(img_rectangle_circle)
plt.title("Rectangle Image")
plt.axis("on")
plt.show()
```

```python
#Task 5
image_zoom = img_rgb.copy()

image_zoom = image_zoom[30:450, 50:500]

plt.figure(figsize=(4,4))
plt.imshow(image_zoom)
plt.title("Picture in rectangle")
plt.show()
```
